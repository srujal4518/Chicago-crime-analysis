```python
df.drop(columns=['x_coordinate', 'y_coordinate', 'latitude', 'longitude'], inplace=True)
```

```python
df["date"]=pd.to_datetime(df["date"],format="mixed",errors="coerce")
```

```python
df["Crime Classification Code"]=pd.to_numeric(df["Crime Classification Code"],errors='coerce')
```

```python
df['updated_on'] = pd.to_datetime(df['updated_on'], errors='coerce')
```

```python
from sqlalchemy import create_engine

# Create the database connection engine
# Replace 'your_password' with your actual MySQL root password
engine = create_engine('mysql+pymysql://root:root@localhost:3306/crimes')

# This pushes the clean dataframe straight into your MySQL table
df.to_sql(name='crime', con=engine, if_exists='replace', index=False)

print("Data pushed directly from Pandas to MySQL successfully!")
```
